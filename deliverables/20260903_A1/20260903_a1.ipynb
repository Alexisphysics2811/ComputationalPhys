### Ex. 1: Temperature

```python
def temp_identifier(T:float=0):
    if T<=0:
        print("Low Temperature")
    elif 0<T and T<30:
        print("Moderate Temperature")
    else:
        print("High Temperature")

T = float(input("Enter Temperature: "))
temp_identifier(T)
```

### Ex. 2: Kinetic Energy

$$T=\frac{1}{2}mv^2$$

```python
kinetic_energy = lambda m,v: 0.5*m*(v**2)

def energy_comparison(E_max:float=10, E:float=1):
    if E > E_max:
        print(f"Higher than E_max: {E_max}")
    elif E < E_max:
        print(f"Lower than E_max: {E_max}")
    else:
        print(f"Energy is equal to E_max: {E_max}")

m = float(input("Enter Mass: "))
v = float(input("Enter Velocity: "))
E_max = float(input("Enter E_max: "))

K = kinetic_energy(m,v)
print(f"Kinetic Energy: {K}")

energy_comparison(E_max, kinetic_energy(m,v))
```

### Ex. 3: Sum of Squares

$$S=\sum_{i=1}^{N}i^2$$

```python
S = lambda n: sum([i**2 for i in range(1,n+1)])

N = int(input("Enter number of terms: "))
print(f"Sum of squares up to {N}: {S(N)}")
```

### Ex. 4: Sequence Convergence

$$x_{n+1}=\frac{1}{2}\left(x_n+\frac{2}{x_n}\right),\qquad x_0=1$$
$$|x_{n+1}-x_n|<10^{-6}$$

```python
def seq_calculate(tol:int=1e-6):
    i = 1
    seq = [1]
    delta = abs(seq[0]-0.5*(seq[0]+(2/seq[0])))
    while delta > tol:
        seq.append(0.5*(seq[i-1]+(2/seq[i-1])))
        delta = abs(seq[i]-seq[i-1])
        i += 1
    print(f"Sequence calculated with {i-1} steps")
    print(f"Delta: {delta}")
    print(f"Last value: {seq[i-1]}")
    return seq

seq_calculate()
```

### Ex. 5: Uniformly Accelerated Motion

$$x(t)=x_0+v_0t+\frac{1}{2}at^2$$

$$x_0=0,\quad v_0=10,\quad a=-9.81$$

```python
def pos(x0:float=0, v0:float=0, a:float=0, t:int=0):
    return x0+(v0*t)+(0.5*a*(t**2))

x0 = float(input("Enter initial position: "))
v0 = float(input("Enter initial velocity: "))
a = float(input("Enter acceleration: "))

print(f"Initial position: {x0}, initial velocity: {v0}, acceleration: {a}")
for t in range(0,10):
    print(f"t = {t}, x({t}) = {pos(x0, v0, a, t)}")
```

### Ex. 6: Maximum Element of an array

```python
def max(arr:list=[]):
    big = arr[0]
    for x in arr:
        if big < x:
            big = x
    return big

max([5, -2, 10, 11, -7, 5])
```
