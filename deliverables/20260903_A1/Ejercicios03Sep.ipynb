## Ejercicio 1. Temperatura
<hr style="border: none; height: 2px; background-color: steelblue;">

Escribe un programa que, dada una temperatura `T` en grados Celsius, determine en qué intervalo se encuentra:

*   Si $T\leq 0$, debe imprimir Temperatura de congelación.
*   Si $0<T<30$, debe imprimir Temperatura moderada.

## Ejercicio 2. Energía cinética
<hr style="border: none; height: 2px; background-color: steelblue;">

La energía cinética de una partícula de masa $m$ y velocidad $v$ está dada por

$$
E_k=\frac{1}{2}mv^2.
$$

Dados los valores de `m`, `v` y una energía de referencia `Emax` por el usuario:


1.   Calcula la energía cinética de la partícula.
2.   Determina si $E_k$ es menor, igual o mayor que $E_{\mathrm{max}}$.
3.   Imprime la energía calculada y el resultado de la comparación.

```python
# Solución
kinetic_energy = lambda m,v: 0.5*m*(v**2)

def energy_comparison(E_max:float=10, E:float=1):
    if E > E_max:
        print(f"Higher than E_max: {E_max}")
    elif E < E_max:
        print(f"Lower than E_max: {E_max}")
    else:
        print(f"Energy is equal to E_max: {E_max}")

m = float(input("Enter Mass: "))
v = float(input("Enter Velocity: "))
E_max = float(input("Enter E_max: "))

K = kinetic_energy(m,v)
print(f"Kinetic Energy: {K}")

energy_comparison(E_max, kinetic_energy(m,v))
```

## Ejercicio 4. Convergencia de una sucesión
<hr style="border: none; height: 2px; background-color: steelblue;">

Considera la sucesión

$$
x_{n+1}=\frac{1}{2}\left(x_n+\frac{2}{x_n}\right),
$$

con

$$
x_0=1.
$$

Utiliza un ciclo `while` para calcular términos sucesivos hasta que se cumpla

$$
|x_{n+1}-x_n|<10^{-6}.
$$

Al finalizar:

*   imprime el valor obtenido;
*   imprime el número de iteraciones realizadas;
*   identifica a qué número conocido se aproxima la sucesión.

```python
# Solución
def seq_calculate(tol:int=1e-6):
    i = 1
    seq = [1]
    delta = abs(seq[0]-0.5*(seq[0]+(2/seq[0])))
    while delta > tol:
        seq.append(0.5*(seq[i-1]+(2/seq[i-1])))
        delta = abs(seq[i]-seq[i-1])
        i += 1
    print(f"Sequence calculated with {i-1} steps")
    print(f"Delta: {delta}")
    print(f"Last value: {seq[i-1]}")

seq_calculate()
```

x(t)=x_0+v_0t+\frac{1}{2}at^2.

Después, utiliza un ciclo for y la función que acabas de escribir para calcular la posición de una partícula entre $t=0\,s$ y $t=10\,s$, cada $\Delta t = t_{j+1} - t_j =  1\,s$

Utiliza:

$$
x_0=0,\qquad
v_0=10\ {\rm m/s},\qquad
a=-9.81\ {\rm m/s^2}.
$$

```python
# Solución
def pos(x0:float=0, v0:float=0, a:float=0, t:int=0):
    return x0+(v0*t)+(0.5*a*(t**2))

x0 = float(input("Enter initial position: "))
v0 = float(input("Enter initial velocity: "))
a = float(input("Enter acceleration: "))

print(f"Initial position: {x0}, initial velocity: {v0}, acceleration: {a}")
for t in range(0,10):
    print(f"t = {t}, x({t}) = {pos(x0, v0, a, t)}")
```

## Ejercicio 6. Elemento máximo de un arreglo
<hr style="border: none; height: 2px; background-color: steelblue;">

Escribe una función `maximo()` que reciba un arreglo `A` y regrese el elemento de mayor valor.

Por ejemplo, para

A = [3, 8, -2, 15, 7, 4]

la función debe regresar

15

Debes recorrer los elementos del arreglo mediante un ciclo.

No utilices `max()`, `np.max()` ni funciones equivalentes.

```python
# Solución
def maximum_ex(arr:list=[]):
    big = arr[0]
    for x in arr:
        if big < x:
            big = x
    return big

maximum_ex([3, 8, -2, 15, 7, 4])
```

## Ejercicio 7. Norma de un vector
<hr style="border: none; height: 2px; background-color: steelblue;">

La norma de un vector

$$
\vec{v}=(v_1,v_2,\ldots,v_n)
$$

está definida como

$$
|\vec{v}|=
\sqrt{
v_1^2+v_2^2+\cdots+v_n^2
}.
$$

Escribe una función `norma(v)` que reciba un arreglo de una dimensión y regrese su norma.

La función debe funcionar para vectores de cualquier tamaño y debe realizar la suma mediante un ciclo `for`.

Prueba tu función con

$$
\vec{v}=(3,4)
$$

y con

$$
\vec{v}=(1,2,3,4).
$$

```python
# Solución
import numpy as np

def norm(v:list=[]):
    S = 0
    for x in v:
        S = S + x**2
    return np.sqrt(S)

print(f"Euclidean norm of v1: {norm([3,4])}")
print(f"Euclidean norm of v2: {norm([1,2,3,4])}")
```

## Ejercicio 9. Construcción de una matriz diagonal
<hr style="border: none; height: 2px; background-color: steelblue;">

Escribe una función `diagonal(n, a)` que construya y regrese una matriz cuadrada de dimensiones $n\times n$ cuyos elementos de la diagonal sean todos`a` y los demás elementos sean cero.

Por ejemplo, *diagonal(3, 5)* debe producir

$$
\left[
\begin{array}{ccc}
5&0&0\\
0&5&0\\
0&0&5
\end{array}
\right]
$$

Comienza creando una matriz de ceros y utiliza uno o más ciclos for para modificar sus elementos.

```python
# Solución
def diag(n, a):
    matrix = [[0 for j in range(n)] for i in range(n)]
    for i in range(n):
        for j in range(n):
            if i == j:
                matrix[i][j] = a
    return matrix

print(f"Diagonal: {diag(3, 5)}")
```

## Ejercicio 10. Jumping game
<hr style="border: none; height: 2px; background-color: steelblue;">

Dada una lista de enteros `jumps` donde cada entrada indica el salto máximo que se puede dar desde esa posición. Escribe una función que determine si es posible llegar desde el primer elemento hasta el último con los saltos disponibles.


```python
# Solución
def jump_game(jumps):
    max_reach = 0
    for i in range(len(jumps)):
        if i > max_reach:
            return False
        if i + jumps[i] > max_reach:
            max_reach = i + jumps[i]
        if max_reach >= len(jumps) - 1:
            return True
    return False

print(jump_game([2, 3, 1, 0, 0]))
```
