### Newton-Raphson

$$f(x)=x^3-10x^2+5,\qquad f'(x)=3x^2-20x$$

$$f(x_1)\approx f(x_0)+f'(x_0)(x_1-x_0)=0$$

$$x_0=1,\qquad f(1)=-4,\qquad f'(1)=-17$$

$$x_1=x_0-\frac{f(x_0)}{f'(x_0)}=1-\frac{-4}{-17}=\frac{13}{17}\approx0.7647058824$$

```python
import pandas as pd

f = lambda x: x**3 - 10*x**2 + 5
f_prime = lambda x: 3*x**2 - 20*x

x0 = 1
x1 = x0 - f(x0)/f_prime(x0)
print('First approximation:', x1)
```

```python
def newton_raphson(f, f_prime, x0, tol=1e-6):
    x1 = x0 - f(x0)/f_prime(x0)
    n = 1

    while abs(x1 - x0) > tol:
        x0 = x1
        x1 = x0 - f(x0)/f_prime(x0)
        n += 1

    return x1, n

print(newton_raphson(f, f_prime, x0))
```

```python
reference = 0.7346035077893032
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = newton_raphson(f, f_prime, x0, tol)
    error = abs(reference - root)/reference*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```
