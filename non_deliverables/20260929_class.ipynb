```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
f = lambda x: x**3 - 10*x**2 + 5
f_prime = lambda x: 3*x**2 - 20*x
ref = 0.7346035077893032
```

```python
def incremental(f,x0,b,h):
    x1 = x0 + h
    n = 1
    while f(x0) > 0 and f(x1) > 0:
        if x1 > b:
            return "None"
        x0 = x1
        x1 = x0 + h
        n+=1
    return x1, n
```

```python
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = incremental(f, 0, 1, tol)
    error = abs(ref - root)/ref*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```

```python
def bisection(f,x0,x1,tol):
    n = 1
    while abs(x0 - x1) > tol:
        x2 = (x0 + x1)/2
        if f(x0)*f(x2) < 0:
            x1 = x2
        if f(x1)*f(x2) < 0:
            x0 = x2
        n+=1
    return x1,n
```

```python
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = bisection(f, 0, 1, tol)
    error = abs(ref - root)/ref*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```

```python
def newton_raphson(f, f_prime, x0, tol=1e-6):
    x1 = x0 - f(x0)/f_prime(x0)
    n = 1
    while abs(x1 - x0) > tol:
        x0 = x1
        x1 = x0 - f(x0)/f_prime(x0)
        n += 1
    return x1, n
```

```python
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = newton_raphson(f, f_prime, 1, tol)
    error = abs(ref - root)/ref*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```

```python
def secant(f, x0, x1, tol):
    n = 1
    while abs(x1 - x0) > tol:
        x2 = x1 - f(x1)*((f(x1) - f(x0))/(x1 - x0))**(-1)
        x0 = x1
        x1 = x2
        n += 1
    return x1, n
```

```python
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = secant(f, 0, 1, tol)
    error = abs(ref - root)/ref*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```

```python
def ridders(f, x0, x1, tol):
    n = 1
    while abs(x0 - x1) > tol:
        x2 = (x0 + x1)/2
        x3 = x2 + (x1 - x0)*((f(x2)/f(x0))/(np.sqrt((f(x2)/f(x0))**2 - (f(x1)/f(x0)))))
        x0 = x2
        x1 = x3
        n += 1
    return x1, n
```

```python
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = ridders(f, 0, 1, tol)
    error = abs(ref - root)/ref*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```
