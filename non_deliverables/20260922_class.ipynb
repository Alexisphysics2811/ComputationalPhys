```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
f = lambda x: x**3 - 10*x**2 + 5
```

```python
def incremental(f,x0,b,h):
    x1 = x0 + h
    n = 1
    while f(x0)>0 and f(x1)>0:
        if x1 > b:
            return "None"
        x0 = x1
        x1 = x0 + h
        n+=1
    return x1, n
```

```python
reference = 0.7346035077893032
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = incremental(f, 0, 1, tol)
    error = abs(reference - root)/reference*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```

```python
def bisection(f,x0,x1,tol):
    n = 1
    while abs(x0-x1)>tol:
        x2 = (x0+x1)/2
        if f(x0)*f(x2)<0:
            x1=x2
        if f(x1)*f(x2)<0:
            x0=x2
        n+=1
    return x1,n
```

```python
reference = 0.7346035077893032
rows = []
tolerances = []

for i in range(1, 6):
    tol = 10**(-i)
    root, steps = bisection(f, 0, 1, tol)
    error = abs(reference - root)/reference*100
    tolerances.append(tol)
    rows.append([root, steps, error])

columns = ['Root', 'Steps', 'Error %']
pd.DataFrame(rows, index=tolerances, columns=columns)
```
