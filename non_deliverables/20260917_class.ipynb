```python
import decimal
from random import *
l = []
for i in range(1000):
    x = uniform(0.8, 2.8)
    l.append(x)

n = len(l)
sum = 0
for i in range(n):
    sum += l[i]
avg = sum / n

error = max(l) - min(l)
print('The resulting average is: \n', avg, '+-', error)
```

```python
def err_abs(theo, approx):
    return abs(theo - approx)

def err_rel(theo, approx):
    return abs(theo - approx) / theo

def err_perc(theo, approx):
    return err_rel(theo, approx) * 100

from math import e
print('Absolute error: %5.4f' %err_abs(e,19/7))
print('Relative error: %5.4f' %err_rel(e,19/7))
print('Percentual error: %7.6f' %err_perc(e,19/7) + '%')
```

```python
from decimal import Decimal as D, getcontext as C
import pandas as pd
from math import sqrt

C().prec = 6

def f(x):
    return x * (sqrt(x + 1) - sqrt(x))
def fd(x):
    return x * (C().sqrt(x + D(1)) - C().sqrt(x))
res = []
idx = []
for i in range(10):
    x = 10**i
    idx.append(x)
    res.append([fd(D(x)), f(x)])
cols = ['Approx', 'Exact']
pd.DataFrame(res, index=idx, columns=cols)
```
