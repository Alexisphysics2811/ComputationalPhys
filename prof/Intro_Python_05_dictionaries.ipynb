# Dictionaries

Un **diccionario** de Python es una estructura de datos mixtos que almacena los mismos en pares llave - valor, donde cada clave es única y se utiliza para recuperar su valor guardado. Se utiliza principalmente cuando se desea almacenar y acceder a datos por nombre en lugar de por posición, como en una lista.

Los diccionarios se pueden crear apartir de la notación de llaves { } o el método **dict()**. Por ejemplo

```python
diccionario1 = {"Nombre": "Alan",    # Notación de llaves
               "Edad" : 22,
               "Carrera": "Física",
               "Universidad": "UNAM"}
```

```python
diccionario2 = dict(Nombre = 'Alan', Edad = 22, Carrera = 'Física', Universidad = 'UNAM')
```

```python
print(diccionario1)
print(diccionario2)
```

Cómo mencionamos en el enunciado anterior, la forma de acceder a los datos almacenados es a través de las llaves en lugar de los índices:

```python
print(diccionario2[0]) # Error al usar la notación de índices
```

La celda anterior nos devuelve un *KeyError* porque no existe ninguna llave que coincida con lo que ingresamos

```python
print(diccionario2['Nombre'])
```

o de igual forma podemos usar el método .get()

```python
diccionario2.get('Nombre')
```

Esta misma notación de llave valor nos permite actualizar y modificar los diccionarios. Por ejemplo

```python
diccionario_frutas = {}  # Llenaremos el diccionario de frutas de la siguiente forma key = 'color' y value = 'lista de frutas'

diccionario_frutas['rojo'] = ['manzana', 'fresa', 'cereza']  # agregamos frutas de color rojo
diccionario_frutas['amarillo'] = ['mango', 'platano']        # agregamos frutas de color amarillo
```

```python
diccionario_frutas
```

Pero ¿qué pasa si en en las frutas rojas he olvidado añadir a la sandía? Podemos actualizar el diccionario de la siguiente manera

```python
diccionario_frutas['rojo'] = ['manzana', 'fresa', 'cereza', 'sandía']
diccionario_frutas # diccionario actualizado
```

También podemos hacerlo usando el método **.update()**

```python
diccionario_frutas = {}  # Llenaremos el diccionario de frutas de la siguiente forma key = 'color' y value = 'lista de frutas'

diccionario_frutas['rojo'] = ['manzana', 'fresa', 'cereza']  # agregamos frutas de color rojo
diccionario_frutas['amarillo'] = ['mango', 'platano']    

diccionario_frutas_rojas = {'rojo':['manzana', 'fresa', 'cereza', 'sandía']}
diccionario_frutas.update(diccionario_frutas_rojas )
```

```python
diccionario_frutas
```

o bien podemos añadir la fruta faltante directamente sobre el valor usando los métodos de listas que ya conocemos

```python
diccionario_frutas['rojo'].append('frambuesa') 
```

```python
diccionario_frutas
```

Al igual que las listas, los diccionarios pueden concatenarse de una forma muy particular

```python
libros1 = {'León Tolstói':'Guerra y Paz', 'George Orwell': '1984'}
libros2 = {'Truman Capote':'A sangre fría', 'Virginia Wool': 'Las olas'}

diccionario_libros = libros1 | libros2
diccionario_libros
```

Otra propiedad importante de los diccionarios es que uno/a puede extraer de forma separada tanto las llaves cómo los valores en formato de lista

```python
autores_as = diccionario_libros.keys() # Extraemos las llaves
print(autores_as)
```

```python
libros = diccionario_libros.values() # Extraemos los valores
print(libros)
```

Los elementos del diccionario se pueden eliminar mediante métodos que funcionan con claves:

- del: elimina un elemento usando su clave
- pop( ): elimina el elemento con la clave dada y devuelve su valor
- clear( ): elimina todos los elementos del diccionario
- popitem( ): elimina y devuelve el último par clave-valor insertado

```python
data = {'Nombre' : 'Alan', 'Edad': 22, 'Nacionalidad':'Mexicana', 'Universidad': 'UNAM'}

del data['Nombre'] # Quitamos la llave 'Nombre' junto con su valor
print(data)

val = data.pop('Edad') # Quitamos la llave 'Edad' y guardamos el valor
print(val)

key, val = data.popitem() # Quitamos la última llave y su valor asociado, pero los guardamos en variables
print(f"Key: {key}, Value: {val}")

data.clear() # Quitamos todo
print(data)
```
