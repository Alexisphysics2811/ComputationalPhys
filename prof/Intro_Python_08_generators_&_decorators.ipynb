# Generators

Los generadores de Python son un tipo especial de función que permite definir una función con un comportamiento iterativo. Ofrecen una forma rápida y limpia de crear secuencias de datos, produciendo valores uno a uno en lugar de devolverlos todos de una vez.

## ¿Cuál es la diferencia entre un generador y una función común y corriente?

Para responder esto, recordemos que una función se ejecuta cuando la llamas y devuelve un valor. Luego de eso, termina sin conservar ningún estado, por lo que la siguiente llamada es totalmente independiente.

Un generador contiene la instrucción `yield`. Al llegar a ella, devuelve un valor pero guarda su estado interno. Esto resulta muy útil para calcular secuencias, como la de Fibonacci: cada llamada puede devolver el siguiente número de la serie gracias a que se conserva dicho estado interno.

### Los dos restaurantes

Una forma didáctica (y bastante divertida) para entender la diferencia entre los generadores y las funciones, consiste en hacer la anaología de ambos con restaurantes 

Imaginemos que en la ciudad existen dos restaurantes rivales de nombre **Func Restaurant** y **Gen Restaurant**

<center>
<img src="restaurants.png" width =750px>
</center>

Ambos restaurantes tienen reglas de operación muy estrictas las cuales no se pueden cambiar y siempre deben seguirse al pie de la letra por el personal de trabajo: 

- La regla de **Func Restaurant** es que, una vez recibida una comanda completa de ordenes del día, se deben elaborar los platillos y entregarse todos a la vez antes de poder hacer cualquier otra cosa.
- La regla de **Gen Restaurant** es que, una vez recibida una comanda completa de ordenes del día, los platillos se entregan uno a uno, en el orden que fueron pedidos, pero esta vez se le permite al personal hacer tareas intermedias (de manera opcioanl) entre cada entrega.

A partir de las descripciones anteriores, podemos llevar este ejemplo al código. Para ello, simplifcaremos un poco las cosas y supondremos que ambos restaurantes comparten el mismo menú.

```python
menú = {0:'Pizza', 1:'Ensalada', 2:'Pan con ajo', 3:'Chuleta', 4:'Crema de elote', 5:'Pambazo', 6:'Hot dog'}
```

```python
def Func_restaurant(comanda): 
    
    entrega = []
    
    for pedido in comanda:

        entrega.append(menú[pedido])

    return entrega
```

```python
def Gen_restaurant(comanda): 

    for pedido in comanda:

        yield menú[pedido]
```

```python
comanda_del_día = [1,0,5,3,4]
```

```python
print(Func_restaurant(comanda_del_día))
```

```python
gen = Gen_restaurant(comanda_del_día)

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))

print('Tarea intermedia!!')

print(next(gen))
```

Cuando uno ve por primera vez ejemplos cómo el anterior es natural que tenga muchas dudas. La primera de ellas podría ser ¿Podríamos cambiar la la palabra `yield` por `return` en `Gen_restaurant()` y obtener el mismo resultado?

La respuesta corta es no, porque de ser así, una vez que retornamos un valor en una función, no retornar ninguna otra. Es importante tener muy claro que `return` es una instrucción final siempre. 

```python
def Func_Gen_restaurant(comanda): 

    for pedido in comanda:

        return menú[pedido]
```

```python
func_gen = Func_Gen_restaurant(comanda_del_día)

print(func_gen)
print(func_gen)
print(func_gen)
print(func_gen)
```

Por más que llamemos a la función, devuelve siempre el primer valor.

A partir de lo que acabamos de revisar, es fácil ver que el uso de las funciones o de los generadores depende mucho de los recursos computacionales que dispongamos y lo que queramos hacer con ellos. 

Siguiendo la analogía de los restaurantes, a la hora de elegir en cual de los dos vamos almorzar, puede que Func Restaurant sea la mejor opción cuando hay poco clientes en la fila y, por el contrario, puede que Gen Restaurant sea la mejor opción cuando hay muchos clientes en la fila.

# Decorators
Un decorador es un tipo especial de función que toma otra función cómo entrada, extiende o modifica su comportamiento y devuelve una nueva función, sin alterar el código de la función original.

La sintaxis es exactamente igual a la de las funciones, sólo que el parámetro de entrada debe ser una función. Veamos el siguiente ejemplo 

```python
def decorador(f): 

    def inner():
        print('Antes de llamar a la función')
        f()
        print('Después de llamara a la función')

    return inner
```

```python
def saludo(): 
    print('Hola!!')
```

```python
decorador(saludo)()
```

```python
def despedida(): 
    print('Adiós!!')
```

```python
decorador(despedida)()
```

Aunque esta forma de llamar a los decoradores es correcta, Python tiene una sintaxis mucho más corta que liga el decorador a la función desde que es declarada. La forma de aplicar esto es mediante `@name_decorator`. 

Por ejemplo

```python
@decorador
def perrito():
    print('woof!')
```

Con esto nos ahorramos la notación y además ligamos de forma permanente el decorador.

```python
perrito()
```

Hasta ahora, todos los decoradores que hemos construido se basan en funciones vacías, es decir, no reciben parámetros de entrada. La pregunta ahora es ¿Cómo podemos implementar los decoradores cuando una función sí recibe parámetros de entrada?

Para esto, usamos la notación de `*args` y `**kwargs`, porque justamente no sabemos cuales serán los parámetros de la función.

Por ejemplo, consideremos el siguiente decorador que mide el tiempo de ejecución de cualquier función.

```python
import time

def medir_tiempo(func):
    def wrapper(*args, **kwargs):
        t0 = time.time()
        result = func(*args, **kwargs)
        tf = time.time()
        dt = tf - t0
        print('La tardó {} segundos en calcular el resultado: '.format(dt))
        return result
    return wrapper
    
@medir_tiempo
def factorial(n): 
    fact = 1
    for i in range(n,1,-1):
        fact *= i
    return fact

result = factorial(5)
print("Resultado:", result)
```
