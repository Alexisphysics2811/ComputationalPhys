# Strings

En Python, un **string** es una secuencia de caracteres entre comillas, la cual puede incluir letras, números, símbolos y espacios.

A diferencia de otros lenguajes de programación, Python no tiene un tipo de carácter independiente (típicamente denominado **char**). Un solo carácter se considera una cadena de longitud uno.

Las cadenas se pueden crear usando comillas simples '...' o dobles "...".


```python
s1 = 'Hola mundo'
s2 = "Hola mundo"

print(s1)
print(s2)
```

Aunque el uso de comillas dobles o simples es indisitinto para la creación de cadenas de caracteres, debe tenerse cuidado con no mezclar ambas instrucciones de la forma incorrecta, ya que puede devolver errores

```python
# Ejemplo del uso incorrecto de comillas dobles y simples
s3 = "El uso incorrecto de comillas para la creación de un 'string" puede devolver errores' 
print(s3)
```

```python
# Ejemplo del uso correcto de comillas dobles y simples
s4 = 'El uso incorrecto de comillas para la creación de un "string" puede devolver errores' 
print(s4)
```

Al igual que las listas y las tuplas, los strings son son secuencias indexadas. Por lo cual, cada uno de los elementos o caracteres que componen a un string cuentan con un índice  que empiezan en 0 desde hasta N-1 (siendo N la longitud del string) de izquierda a derecha; y en -1 de derecha a izquierda

```python
s5 = "Hola mundo"
print(s5[0])
print(s5[1])
print(s5[2])
print(s5[3])
```

Algo interesante y que, por desgracia no siempre es mencionado, es que los strings de python se codifican a partir del ASCII ( por sus siglas en inglés American Standard Code for Information Interchange). Por lo cual, todos los caracteres que se usan para la creación de los strings tienen un número de identificación dentro de la base de datos de ASCII. **Importante** este número de indentifiación es totalmente diferente al **index** del caracter en la cadena que se construya, por lo cual no deben confundirse. 

Para obtener el valor o número de identificación ASCII (punto de código Unicode) de **un solo carácter**, utilizamos la función ord().

```python
c = 'a'
print(ord(c))
```

```python
s_prueba = "Hola mundo!"

print('Caracter:', s_prueba[0],'-> Valor ASCII:', ord(s_prueba[0]), '-> Índice de lista:', 0)
print('Caracter:', s_prueba[1],'-> Valor ASCII:', ord(s_prueba[1]), '-> Índice de lista:', 1)
print('Caracter:', s_prueba[2],'-> Valor ASCII:', ord(s_prueba[2]), '-> Índice de lista:', 2)
print('Caracter:', s_prueba[3],'-> Valor ASCII:', ord(s_prueba[3]), '-> Índice de lista:', 3)
print('Caracter:', s_prueba[4],'-> Valor ASCII:', ord(s_prueba[4]), '-> Índice de lista:', 4)
print('Caracter:', s_prueba[5],'-> Valor ASCII:', ord(s_prueba[5]), '-> Índice de lista:', 5)
print('Caracter:', s_prueba[6],'-> Valor ASCII:', ord(s_prueba[6]), '-> Índice de lista:', 6)
print('Caracter:', s_prueba[7],'-> Valor ASCII:', ord(s_prueba[7]), '-> Índice de lista:', 7)
print('Caracter:', s_prueba[8],'-> Valor ASCII:', ord(s_prueba[8]), '-> Índice de lista:', 8)
print('Caracter:', s_prueba[9],'-> Valor ASCII:', ord(s_prueba[9]), '-> Índice de lista:', 9)
print('Caracter:', s_prueba[10],'-> Valor ASCII:', ord(s_prueba[10]), '-> Índice de lista:', 10)
```

Y así cómo podemos obtener el valor de un caracter en el código ASCII, podemos realizar el proceso inverso

```python
value = 505
print(chr(value)) 
```

Los strings, a diferencia de las listas, son inmutables, lo que significa que no se pueden modificar una vez creadas:

```python
string_prueba = 'python'
string_prueba[0] = 'P'
```

```python
string_prueba = 'python'
```

```python
str_new = 'python' + ' cool'
str_new
```

Aunque los strings no se pueden modificar **inplace** como las listas, existen algunos métodos y trucos para modificarlos a partir de la concatenación de strings y la redeclaración de variables

```python
string_prueba = 'python'

print('string original:', string_prueba)

string_prueba = 'P' + string_prueba[-5:] # Uso de concatenación y slicing igual que en las listas
print('string nuevo:', string_prueba)
```

```python
string_a = 'a'
string_a4 = string_a*4
print(string_a4)
```

### Métodos de utilidad en strings

Python proporciona varios métodos integrados para manipular cadenas. Algunos de ellos son: 

- upper(): convierte todos los caracteres a mayúsculas.

- lower(): convierte todos los caracteres a minúsculas.

- capitalize(): convierte en mayusculas únicamente la primera letra y el resto las convierte en minúsculas.
  
- strip(): elimina los espacios iniciales y finales de la cadena.

- replace(): reemplaza todas las apariciones de una subcadena especificada con otra.

```python
s = "Hola mundo"
print(s.upper())
print(s.lower())
print(s.capitalize())
```

```python
s = "   Python   "
print(s.strip())    

s = "Python es aburrido"
print(s.replace("o", "0"))
```

## Formato de cadenas

Python ofrece varias maneras de incluir variables dentro de cadenas, lo cual es útil para cuando queremos mostrar valores guardados en variables.

```python
nombre = "Ana"
edad = 24
print(f"Nombre de usuario: {nombre}, Edad: {edad}")
```

Otra forma de hacer esto es utilizando el método format():

```python
nombre = "Ana"
edad = 24
print("Nombre de usuario: {}, Edad: {}".format(nombre, edad))  # las variables/valores dentro de format() se imprimen en el mismo orden que los espacios vacíos {}
```

Python permite especificar el tipo de formato que tendrán los strings , como enteros, números de coma flotante, etc. Esto resulta útil para controlar la visualización de valores numéricos, fechas y la alineación del texto.

```python
s1, s2, s3 = "Física", "Ciencias","Python"

print("{:<20} es genial.".format(s3))  # Añade 20 espacios después de insertar el string
print("En la facultad de {} se estudia la licenciatura: {:.3}".format(s2, s1)) # trunca el string
```

```python
print("El número pi es aproximadamente igual a {:.8f}".format(3.14159265358979323846))
```
