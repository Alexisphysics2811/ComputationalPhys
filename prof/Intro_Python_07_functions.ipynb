# Functions

En Python, una función es un bloque de código definido para realizar una tarea específica. Las funciones se utilizan para encapsular acciones comunes o repetitivas dentro de un programa. De esta forma, en lugar de escribir el mismo código varias veces para distintas entradas, es posible invocar (o llamar) la función y reutilizar el código que contiene.

Una manera útil de interpretar las funciones en Python es considerarlas como cajas negras que reciben una cierta cantidad de variables o parámetros (o ninguno), para luego ejecutar una serie de instrucciones con ellos y, en general, devolver un resultado.

En python, a diferencia de otros lenguajes, no es necesario especificar el tipo de variable que devolverá nuestra función ya que esto viene implicito al usar las sentencias reservada **def** y **return**. La primer sentencia se emplea para "definir" que la variable *funciton_name* es propiamente una función, mientras que la segunda se emplea para devolver la salida de lo que sea que haga nuestra función, de ser ese el caso. 

Aunque no siempre es necesario que una función "regrese algo" (a este tipo de funciones se les conoce como tipo *void* en lenguajes coomo C/C++, Java, etc), es recomendable no olvidar programar la setencia **return** antes que el cuerpo mismo del código. 

Para comenzar, revisemos dos funciones sencillas con y sin **return**

```python
def print_mayusculas_no_return(x):    # Declaramos la función con x el string de entrada

    x = x.upper() + "!!!"   # redefinimos a x para que aparezca en mayusculas y agregamos !!!
    print(x)                # imprimimos el resultado
                            # no usamos return
```

```python
print_mayusculas_no_return('hola')
```

```python
def print_mayusculas(x):    # x es el string de entrada

    x = x.upper() + "!!!"   # redefinimos a x para que aparezca en mayusculas y agregamos !!!
    return x                # usamos return
```

```python
print_mayusculas('hola')    # invocamos la función 
```

Cómo se observa, ambas funciones hacen exactamente lo mismo, con una sutil diferencia. Veamos lo siguiente

```python
x = print_mayusculas_no_return('hola')  # Guardamos el resultado de ambas funciones en variables
y = print_mayusculas('hola')
```

```python
print(type(x))  # AL imprimir el tipo de variable, vemos que la primera función nos devuelve un dato sin tipo definido
print(type(y))  # mientras que la segunda nos devuelve un string
```

Si quisieramos usar el resultado de la primera función para digamos... crear un nuevo string, no podríamos: 

```python
new_string = x + ' :)'
print(new_string)
```

Porque nunca podremos hacer ningún tipo de operación o modificación sobre algo que no tiene un tipo de dato bien definido. 

Por otro lado, si hacemos lo mismo con la otra variable...

```python
new_string = y + ' :)'
print(new_string)
```

El resultado es disntinto y satisfactorio.

Continuando con nuestra discusión de las funciones, previamente mencionamos que estas pueden recibir uno, dos, tres o más parámetros, o bien, pueden no recibir ninguno. Esto no representa ningún problema ya que no todas las funciones necesitan de un parámetro para que sus instrucciones se lleven a cabo, por ejemplo

```python
def uno():   # no recibe ningún parámetro

    return 1 # regresa 1 
```

Cómo pudimos observar, no representa ningún error el hecho de que la función no reciba parámetro alguno. Lo que sí puede llevar a un error, es proporcionarle un parámetro cuando la función no lo necesita:

## Valores predefinidos en los parámetros de entrada

Pero **¿qué pasa en caso de que sea el usuario quien deba ingresar los parámetros a la función y por descuido olvide ingresarlos todos?**

Para tal caso, existe una alternativa que podemos usar para prevenir el error humano, esto es, predefinir los parámetros dentro de la función con valores específicos usando `=`. Esto es, la una declaración interna de las variables dentro de los paréntesis: 

```python
def potencia(a=1,x=1): 

    return a**x
```

```python
potencia(2)
```

```python
a = input('Ingrese un número: ')
x = input('Ingrese una potencia: ')

print(potencia(float(a), float(x)))
```

Aunque dentro de la función las variables tengan un orden definido, este orden pueden pasarse por alto al invocar la función siempre y cuando se especifiquen los nombres de las variables que vamos a proporcionar:

```python
print(potencia(a = 5, x = 2))
print(potencia(x = 2, a = 5))
```

En python, las funciones no sólo pueden recribir como parámetros a variables "simples", en realidad una función también puede recibir alguna estructura de datos como parámetro y realizar todo tipo de acciones sobre ella. Por ejemplo: 

```python
def itera_lista(lista): 

    for elemento in lista: 
        
        print(elemento)
    print('Fin!')
```

```python
itera_lista([1,2,3])
```

Dado el código anterior, una pregunta interesante es... dentro de una función ¿se puede iterar sobre las variables de entrada? 

Al principio, esta pregunta parece bastante inocente ya que hay dos soluciones rápidas y obvias: 
1. Si pedimos que el único parámetro sea una estructura de datos que contenga N parámetros de entrada al igual que antes, directamente podemos iterar sobre ellas.
3. Si recibimos N variables de entrada por separado, podemos guardarlas en una lista y luego iterar sobre esta:
   

```python
def itera_params_1(lista): # Solución 1.
    
    for elemento in lista: 
        print(elemento)
        
    print('Fin!')

def itera_params_2(x,y,z): 

    lista = list((x,y,z)) 
    for elemento in lista:
        print(elemento)
    print('Fin!')
```

```python
print('Resultado solución 1. : ')
itera_params_1([1, 2, 3])
print('Resultado solución 2. : ')
itera_params_2(1, 2, 3)
```

Aunque estas soluciones son correctas, tienen algunas sutilezas que podrían dar problemas. Por ejemplo, la solución 1. requiere que, de manera forzosa, se le proporcione a la función una estructura de datos finita, ya que si recibe una variable que no cumpla esto, la función mostrará error

```python
itera_params_1(1)
```

Por otro lado, la solución 2. requiere que de forma manual guardemos los parámetros de entrada en la lista, esto en principio no representa un problema cuando tenemos pocas variables, pero si el número de variables es lo suficientemente grande entonces la solución es impracticable.

Ante estas problemáticas existe una tercera solución que nos permite ingresar tantas variables cómo querramos sin tener que indicarles nombres y sobre las cuales podremos iterar sin tener que almacenarlas en una lista. Esto se especifica mediante:`*args` y `**kwargs`.

- `*args` se utiliza cuando se sabe cuántos argumentos posicionales podría pasar un usuario a la función.
- `**kwargs` se utiliza cuando se quieren manejar argumentos con llave y valor que no hayan sido predefinidos. Similar a los diccionarios.

Por ejemplo

```python
def itera_params_3(*args):

    for param in args:
        print(param)
    print('Fin')
```

```python
print('Resultado solución 3. : ')
itera_params_3(1,2,3)
```

## Tipo predefinido en parámetros de entrada

Así cómo podemos predefinir los valores de los parámetros de entrada, también podemos predefinir el tipo al cual pertenecerá la variable. Esto se hace mediante el uso de `:`. Al combinar esto con los valores predefinidos se pueden construir funciones mucho más estructuradas y limpias que nos ayudan a mejorar el enfoque de nuestro problema a resolver.

### Documentation 

En cualquier lenguaje de programación, una buena práctica es escribir la documentación de las funciones complicadas para que otras personas puedan entender qué es lo que hace y leer algunas otras especificaciones que puedan ser de útilidad en caso de que se esté haciendo un uso incorrecto de la misma. 

Para escribir la documentación de una función se emplean tres comillas simples ''' al principio y final de la documentación, por ejemplo:

```python
def potencia(x : int = 1, y : int = 1):

    '''
    Función que calcula el resultado de elevar el número x a la potencia y. 

    En caso de que no se le proporcionen los argumentos o parámetros a la función, esta considera los valores predefinidos.

    Parámetros:  
        x: int, float, complex, default: 1
            número base
            
        y: int, float, complex, default: 1
            potencia
    
    Returns:

        x**y : int, float, complex
    '''
    return x**y
```

La documentación nunca será visible cuando se invoque la función

```python
potencia(2,2) 
```

Para que alguien pueda consultar la documentación de nuestras funciones sin tener que abrir el código fuente, basta con usar help(...)

```python
help(potencia)
```

## Lambda

Las funciones lambda son pequeñas funciones anónimas, lo que significa que no tienen un nombre definido. En Python, las funciones lambda se crean usando la palabra reservada **lambda** para operaciones cortas y sencillas. 

Las funciones lambda no necesitan **return** ya que este está implicito (sobre la misma linea) posterior a los dos puntos

```python
potencia = lambda x,y: x**y
potencia(2,2)
```

En las funciones lambda también podemos predefinir los valores de los parámetros

```python
potencia = lambda x=1,y=1: x**y
potencia()
```

```python
check = lambda x: "Positivo" if x > 0 else "Negativo" if x < 0 else "Cero"
print(check(5))   
print(check(-3))  
print(check(0))
```

## Recursion

La recursión es una técnica de programación en la que una función se llama a sí misma, directa o indirectamente, para resolver un problema dividiéndolo en subproblemas más pequeños.

La estructura general de una función recursiva consiste en:

1. Recibir un parámetro del cual queremos obtener un resultado
2. Declarar un caso base con un resultado bien definido
3. Declarar el resto de los casos (denomidados casos recursivos) cuyo resultado se obtiene a partir del llamado de la función misma con el parámetro inicial modificado

La idea general de la recursión consiste en que, a partir del caso base, se puedan resolver los casos subsecuentes. Lo más importante aquí es que, a partir de la modificación del parámetro inicial, podamos llegar al caso base para luego regresar a la solución completa.

Pensar a las funciones recursivas como muñecas matrioskas es de hecho una idea bastante ilustrativa, ya que muestran perfectamente cómo el producto final se construye a partir de productos más pequeños ligeramente diferentes que parten de una base. 

Ejemplo

```python
def factorial(n):
    if n == 0 or n == 1:  # Caso base
        return 1
    else:                 # Caso recursivo
        return n * factorial(n - 1)

print(factorial(5))
```

La función factorial es un gran ejemplo de recursión ya que su definición matemática formal es recursiva en sí. Recordemos que se define el factorial de un número entero positivo cómo:

$$
n! = n(n-1)! = n(n-1)(n-2)! = ... = n(n-1)(n-2)(n-3)...1
$$
