# Operaciones entre variables de tipo numérico

Las operacciones base entre los datos de tipo numérico son aquellas que se establecen dentro de los campos de los números reales y con las que estamos familiarizados desde cálculo I y álgebra superior (o álgebra para físicos). La sintaxis de estas operaciones es muy intuitiva, pero en ocasiones puede jugarnos en contra si no pondemos atención a los detalles

### Suma

$$
a,b \in \mathbb{R}, \mathbb{C} \quad\Rightarrow\quad a + b = b + a = c \in \mathbb{R}, \mathbb{C}
$$

```python
a = 1
b = 2
c = a+b
print(c)
```

```python
a = 1 + 1j
b = 2 + 2j
c = a+b
print(c)
```

### Resta

```python
a = 1
b = 2
c = a-b
print(c)
```

```python
a = 1
b = 2
c = b - a
print(c)
```

```python
c1 = abs(a-b) # valor absoluto
c2 = abs(b-a)

print('|a-b| =',c1,'= |b-a| =', c2)
```

### Multiplicación

$$
a,b \in \mathbb{R}, \mathbb{C} \quad\Rightarrow\quad a * b = b * a = c \in \mathbb{R}, \mathbb{C}
$$

```python
a = 1
b = 2
c = a*b
print(c) 
```

```python
a = 1 + 1j
b = 2 + 2j
c = a*b
print(c)
```

### Potenciación

```python
b = 2
c = b*b*b*b*b
print(c)
```

```python
b = 2
c = b**5
print(c)
```

### División

```python
a = 1
b = 2
c = a/b
print(c)
```

```python
a = 1
b = 2
c = b/a
print(c)
```

### División entera

```python
d = 7
b = 2

c = d/b  # División "normal"
print(c)
```

```python
d = 7
b = 2

c = d//b  # División entera
print(c)
```

### Módulo o Residuo

```python
d = 7
b = 2

c = d%b
print(c)
```

```python
f = d + 1
b = 2

c = f%b
print(c)
```

## El orden de las operaciones importa, por ejemplo...

```python
x = 1+5//2*6*2/4**0
print(x)
```

no es lo mismo que...

```python
x = ( (1 + 5) // 2 ) * 6 * ( (2/4)**0 )
print(x)
```

Los paréntesis nos ayudan a delimitar las operaciones y modifican la jerarquí de las mismas. Python sigue la convención matemática estándar:
