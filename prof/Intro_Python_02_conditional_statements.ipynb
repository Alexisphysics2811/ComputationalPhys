# Condicionales, operadores lógicos y comparativos

Una sentencia condicional se puede interpretar como un filtro que determina la ejecución de una instrucción a partir del cumplimiento de una condición lógica. Este tipo de sentencias se utilizan en la vida diaria aunque muchas veces no nos detengamos a pensar en las implicaciones lógicas que hay de por medio.

Consideremos la siguientes oración de ejemplo: 

<center>
    "Si hace frío, usaré ropa cálida"
</center>


En el ejemplo anterior nuestra condición lógica es que, **si** el clima del presente registra temperaturas bajas, **entonces** es motivo suficiente para que ejecutemos la acción/decisión de usar ropa que nos proteja térmicamente.

En python esa lógica es exactamente la misma, pero no siempre las condiciones tan directas ya que en ocasiones existen subcasos o casos diferentes a nuestras condiciones que nos pueden llevar a ejecutar una o más acciones, tal cómo pasa en la vida real.

## Operadores comparativos

A partir de lo que mencionamos antes, entendemos que si algo es **verdadero** o **falso** se puede o no ejecutar una acción. Cómo ya sabemos, en python existen los datos booleanos **True** y **False**, la cuestión ahora es cómo establecer esos valores. Para eso tenemos los siguientes operadores de comparación:

- mayor o menor que: <,>
- igual a: ==
- mayor igual o menor igual que: <=, >=
- distintio a: !=

Por ejemplo

```python
print(0 < 2) # Sabemos que es cierto que 0 es menor que 2
```

```python
print(10 > 100) # Sabemos que es falso que 10 es mayor a 100
```

```python
print('1' == 1) # El caracter '1' es diferente del valor numérico 1
```

```python
a = 5
print('a' != a) # El valor almacenado en la variable a es diferente al caracter 'a'
```

```python
lista = ['a','b','c','d']
x = 'a'
print(lista == x) # Las variables de tipo diferente no pueden ser iguales
```

De igual forma, existen operadores que indican la pertenencia a alguna secuencia de datos

```python
a = 0
lista_num = [0,1,2,3,4,5]

print(a in lista_num)
```

```python
x = 'x'
print(x not in lista_num)
```

Ya que aprendimos cómo funcionan los operadores de comparación y pertenencia, podemos abordar las sentencias condicionales

## Sentencia if

La sentencia condicional más básica en Python es la sentencia if, esta sentencia se lee cómo "si esto es verdaderdor, entonces..." Todas las líneas de código dentro del sub-bloque definido por la identeación se cumplen siempre que la condición sea verdadera.

```python
if False:          # Si algo es verdadero
                  # Una identación es igual a 4 espacios o un tab
    print('Hola') # imprime la palabra 'Hola'
    
```

```python
x = 0 

if (x != 10):
    
    print('x =', x, 'es menor que 10')
```

```python
fruta = 'manzana' 

if 'a' in fruta:
    
    print('La fruta', fruta, 'tiene al menos una a')
```

## Sentencia else y elif

En algunas ocasiones hay condicionales que no se cumplen bajo una regla estricta, pero sí bajo otra y no por eso dejan de ser interesantes. Por ejemplo, en el ejemplo de la variable x, nos gustaría que python nos indicara cuándo es mayor que 10 y cuándo no. 

Para eso, usamos la sentencia **else**, que siempre debe ir acompañada de un **if** y se lee cómo "si esto es verdadero, entonces..., caso contrario, entonces..."

```python
x = 11                                  # Declaramos la varible

if x < 10:                             # Si x es menor que 10...
    print('x =', x, 'es menor que 10') # imprime la oración

else:                                  # caso contrario (no es necesario indicar la condición ya que es complementaria)
    print('x =', x, 'es mayor o igual que 10')
```

podemos hacerlo más interesante, usaremos una función que permita ingresar el valor que queremos para x a través del teclado

```python
x = input('Ingrese el valor para x: ') # Ingresamos el valor de x mediante el teclado
x = float(x)                           # redefinimos el valor de x para que sea flotante y no string

if x < 10:                             # Si x es menor que 10...
    print('x =', x, 'es menor que 10') # imprime la oración

else:                                  # caso contrario (no es necesario indicar la condición ya que es complementaria)
    print('x =', x, 'es mayor o igual que 10')
```

¿Qué pasa si queremos analizar los mismos casos de antes pero también queremos saber si x es igual a un valor en específico? Digamos un valor guardado en alguna variable...

Para eso usamos la sentencia **elif** que nos permite revisar condiciones que sean tangenciales a nuestras condiciones principales dentro del **if** y **else**

```python
a = 12                                 # Guardamos un valor especial de una variable que sea mayor a 10 

x = input('Ingrese el valor para x: ') # Ingresamos el valor de x mediante el teclado
x = float(x)                           # redefinimos el valor de x para que sea flotante y no string

if x < 10:                             # Si x es menor que 10...
    print('x =', x, 'es menor que 10') # imprime la oración

elif x == a:
    print('x =', x, 'es mayor o igual que 10 y además es igual al valor guardado en a')

else:                                  # caso contrario (no es necesario indicar la condición ya que es complementaria)
    print('x =', x, 'es mayor o igual que 10')
```

## Operadores lógicos

Los operadores lógicos en Python: **and, or, not** permiten combinar o invertir condiciones booleanas, fundamentales para la toma de decisiones en las sentencias. **and** requiere que ambos lados sean ciertos, **or** requiere al menos uno, y **not** invierte el valor cómo vimos antes

```python
a = 16                                              # Guardamos un valor especial de una variable que sea mayor a 10 

x = input('Ingrese el valor para x: ')               # Ingresamos el valor de x mediante el teclado
x = float(x)                                         # redefinimos el valor de x para que sea flotante y no string


if x < 10:                                            # Si x es menor que 10...
    print('x =', x, 'es menor que 10')                # imprime la oración

elif 10 < x  and x < 15:
    print('x =', x, 'está entre 10 y 15')

elif x == a:
    print('x =', x, 'es igual al valor guardado en a')

else:                                                # caso contrario (no es necesario indicar la condición ya que es complementaria)
    print('x =', x, 'es mayor o igual que 15')
```
