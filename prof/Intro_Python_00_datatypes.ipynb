# Tipos de datos

Python cuenta con varios tipos de datos integrados que categorizan los datos y determinan cómo se pueden almacenar y manipular. En Python no es necesario declarar explícitamente el tipo de una variable; el intérprete asigna automáticamente el **tipo en función del valor**.

Los tipos de datos son:

## **Booleanos**

Datos integrado fundamental que representa uno de dos valores posibles: Verdadero o Falso. Cruciales para la lógica y el control de flujo del programa

```python
True
```

```python
False
```

La función (aún no sabemos exactamente qué es una función pero más adelante lo descubriremos) **type()** nos permite identificar de qué tipo es una cosa. Por ejemplo

```python
type(True)
```

## **Numérico**

- int: Números que, en nuestro contexto matemático, conocemos cómo el conjunto de números enteros.
- float: Números con punto decimal.
- complex: Números complejos, es decir, aquellos que se constituyen de una parte real y una imaginaria.

```python
# int
1
```

```python
# float
3.141516
```

```python
# float con notación científica
3e8 # 3x10^8
```

La 'e' o 'E' en la notación científica de python es el equivalente al x10 a la

```python
5E-64 # Cuando el número de decimales es muy grande imprime la notación directa
```

```python
# complex
5 + 4j
```

```python
complex(5,4)
```

```python
type(1)
```

## **Secuencias**

Cómo su nombre lo indica, son secuencias de datos individuales con un orden establecido por índices (o etiquetas) que van desde el 0 hasta n-1, con n = la longitud de la secuencia. 

- str: Secuencia inmutable de **caracteres Unicode** que se define con comillas
- list: Secuencia mutable de datos mixtos que se declara mediante corchetes: [ ]
- tuple: Secuencia inmmutable de datos mixtos que se declara mediante corchetes paréntesis: ( )
- range: Secuencia inmutable de números enteros con incremento fijo que se declara mediante: range(número inicial, número final, incremento)
- otros

```python
# str
'Hola'
```

```python
# list
[1,'a', True]
```

```python
# tuple
(2, 'b', False)
```

## **Mapping**

- dict: Colección de datos mixtos desordenada de pares clave-valor. Se declara mediante llaves { }

```python
{'a': 1, 'lista': ['cosa1', 'cosa2'],'25':25, 25:'veinticinco'}
```

**None**

Datos que no entran dentro de ninguna de las categorías antes mencionadas

# ¿En dónde guardamos estos datos? 

```python
x1 = 0
print(x1) # imprime en pantalla lo que esté guardado en la variable 'x'
```

```python
type(x1)
```

```python
id(x1)
```

```python
x = 'a'
y = 'b'
```

```python
print(id(x)) # Diferentes valores -> diferente número de identificación
print(id(y))
```

```python
x = y
```

```python
print(x)
```

```python
print(id(x)) # x = y ^ y = 1 -> x = 1 ^ id(x) = id(y)
print(id(y))
```

Es importante destacar que tanto el nombre de la variable es único y en caso de que existan dos varibles con el mismo nombre, pero diferente valor, python "elige" el último valor declarado:

```python
x = 1
x = 'uno'
print(x)
```

```python
a = '101010101'
A = '1'

print('Valores iniciales de las variables:')
print('a = ', a)
print('A = ', A, '\n')
```

```python
A*10
```

```python
A = a

print('Reasignación de variables:')
print('a = ', a)
print('A = ', A)
```

```python
A*10
```
