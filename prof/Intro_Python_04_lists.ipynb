# Listas

Las **listas** en Python son colecciones ordenadas y mutables, que pueden contener datos de tipos mixtos y son fundamentales para agrupar información y acceder a cada ítem por su índice. 

<center>
    <img src = 'https://solocodigoweb.com/wp-content/uploads/2017/05/operaciones_con_listas_python.jpg'  width = 500px>
</center>

Las listas se pueden construir de dos maneras principales: 1. usando la notación de corchetes [...] o 2. usando el constructor list(). Por ejemplo

```python
frutas = ['manzana', 'mango', 'fresa', 'platano', 'sandía']
números = [1,2,3,4,5,6,7,8]
abecedario = ['a', 'b', 'c', 'd']

print(frutas)
print(números)
print(abecedario)
```

```python
frutas = list(('manzana', 'mango', 'fresa', 'platano', 'sandía'))
números = list((1,2,3,4,5,6,7,8))
abecedario = list('abcd')

print(frutas)
print(números)
print(abecedario)
```

## Indexing

Los elementos dentro de una lista tienen una etiqueta o índice bien definido que establece su orden, dada una lista de longitud N (i.e. con N elementos), los índices de sus elementos irán desde 0 hasta N-1 por convención. 

<center>
    <img src = 'https://www.alphacodingskills.com/python/img/python-list.png'>
</center>

El índice es de gran utilidad ya que nos permite saber qué estamos guardando en la j-ésima posición de nuestra lista:

```python
frutas = ['manzana', 'mango', 'fresa', 'platano', 'sandía']
print(frutas[2])
```

Los índices pueden leerse de izquierda a derecha, pero también pueden leerse en sentido inverso empleando índices negativos; lo cual es útil cuando tenemos listas de una gran longitud

```python
frutas = ['manzana', 'mango', 'fresa', 'platano', 'sandía']
N = len(frutas)   # Obtenemos la longitud de la lista

print('La lista de frutas tiene N =', N, 'elementos')

print('Elemento con índice N-1:', frutas[N-1])
print('Elemento con índice -1:', frutas[-1])
```

Ya vimos entonces que, si conocemos el índice, podemos conocer el elemento asociado al mismo. Pero ¿qué pasa si estamos sabemos que dentro de nuestra lista hay un elemento bien conocido, pero no sabemos su índice? ¿Cómo podemos obtener el índice sin tener que recurrir a buscar en cada una de las posiciones?

Respuesta: usando el método .index()

```python
frutas = ['manzana', 'mango', 'fresa', 'platano', 'sandía']
x = 'mango'  # elemento conocido

print(x, 'tiene el índice :', frutas.index(x))
```

Los índices en python no sólo nos permiten identificar elementos individuales dentro de nuestra lista, también nos permiten hacer algo conocido cómo **slicing**, esto es, una forma eficaz de extraer porciones (o "segmentos") de la lista.

La sintaxis de segmentación incluye hasta tres parámetros opcionales separados por dos puntos: [inicio:fin:paso]

```python
print('Hacemos slicing sobre toda la lista:',frutas[:])
print('Hacemos slicing sobre una parte de la lista:',frutas[1:4])
print('Hacemos slicing sobre toda la lista en saltos de 2 en 2:',frutas[::2])
```

## Actualización de elementos en listas

Dado que las listas son mutables, podemos actualizar elementos accediendo a ellos a través de su índice. Esto es útil para cuando hemos cometido un error al construir la lista, pero no queremos redefinirla desde cero. Por ejemplo

```python
números = list((1,2,'tres',4,5,6,7,8))
print('Antes:',números)

números[2] = 3   # actualizamos el valor usando el índice
print('Después:',números)
```

## Concatenación de listas

Una propiedad interesante de las listas es que estas pueden **concatenarse**. En un sentido coloquial, si pensamos a las listas cómo un archivero, entonces, concatenar una lista con otra es como si pegaramos dos archiveros entre sí para que formen uno más grande

<center>
    <img src = 'https://st.depositphotos.com/5579928/58405/v/450/depositphotos_584056914-stock-illustration-portrait-employees-office-business-woman.jpg' width = 450>
</center>

por ejemplo

```python
frutas_rojas = ['manzana', 'fresa', 'sandía', 'cereza']
frutas_verdes = ['uva', 'pera', 'kiwi', 'manzana']

frutas = frutas_rojas + frutas_verdes # concatenamos las listas

print(frutas)
```

Cómo se puede observar, la manzana aparece dos veces en la lista de frutas mas grandes. Aunque estricatemente este no es un error ya que existen tanto manzanas verdes como rojas, sí representa un problema de duplicación de datos dentro de nuestra lista ya que no se especifica ninguna diferencia. 

En conclusión, la concatenación de datos sólo sirve para "pegar" listas sin filtrar la duplicación de datos. Tanto es así, que podemos duplicar los elementos de una lista de un sólo elemento tantas veces cómo querramos, de la siguiente forma:

```python
lista_uva = ['uva']

lista_varias_uvas = ['uva']*12

print(lista_varias_uvas)
```

Una vez identificado lo anterior, vale la pena introducir un método que nos permita contar cuantas veces aparece un elemento bien conocido dentro de una lista, sin tener que hacer el conteo en cada índice: 

```python
print(frutas.count('manzana'))
```

## Agregar elementos a listas

Podemos añadir elementos a una lista mediante los siguientes métodos:

- append(): Añade un elemento al final de la lista.

- extend(): Añade varios elementos al final de la lista.

- insert(): Añade un elemento en una posición específica.

Estos métodos funcionan mediante la instrucción **inplace**, es decir, no es necesario redefinir la lista para que se realicen los cambios

```python
abecedario = list('abcd')

abecedario.append('e')
print('Usamos append() para añadir la letra "e" :', abecedario)

abecedario.extend(['f', 'h'])
print('Usamos extend() para añadir las letras "f", "g" :', abecedario)

abecedario.insert(-1, 'g')
print('Usamos insert() para insertar la letra "h" en la penúltima posición :', abecedario)
```

## Eliminar elmentos de la lista

Podemos eliminar elementos de una lista usando:

-remove(): Elimina la primera ocurrencia de un elemento leyendo de izquierda a derecha.

-pop(): Elimina el elemento en un índice específico o el último elemento si no se especifica ningún índice.

-del: Elimina un elemento en un índice específico.

-clear(): Elimina todos los elementos

```python
abecedario = list('abcdefgh')

abecedario.remove('e')
print('Usamos remove() para eliminar la letra "e":', abecedario)

abecedario.pop(-2)
abecedario.pop()
print('Usamos pop() para eliminar las letras "g", "h":', abecedario)

del abecedario[abecedario.index('f')]
print('Usamos del para eliminar la letra "f":', abecedario)

abecedario.clear()
print('Usamos clear() para eliminar todo:', abecedario)
```

## Ordenar listas

En muchas ocasiones, las listas se construyen sin tener un orden bien definido. Por ejemplo, en nuestra lista de frutas los elementos sí tienen sus índices bien definidos de 0 a N-1, pero ¿qué pasaría si quisieramos que los elementos se ordenaran alfabéticamente? Para eso usamos **sort()**

```python
print(frutas)
```

```python
frutas.sort()
print(frutas)
```

¿y si ahora quisieramos ordenar las frutas en orden alfabético inverso y sin que aparezca dos veces la manzana? 

```python
frutas.sort()           # Ordenamos
frutas.reverse()        # Invertimos
frutas.remove('manzana')# Eliminamos el duplicado

print(frutas)
```

## Convertir una cadena de caracteres en una lista

Previamente ya vimos que podemos convertir una lista a partir de un string de la siguiente manera:

```python
list('abcdfgh')
```

¿Pero qué pasa si yo quiero convertir la siguiente oración en una lista?

```python
list('Hola a tod@s')
```

Vemos que, esencialmente, sí almacena todos los elementos del string en lugares de la lista pero... ¿Cómo puedo hacer para que, en vez de guardar cada letra, guarde las palabras sin considerar a los espacios? 

Para eso usamos el método **split()**

```python
lista =  'Hola a tod@s'.split()
lista
```

Lo grandioso de split es que no sólo permite hacer la separación a partir de los espacios, sino que nos permite hacerlo a partir de cualquier delimitador que inpongamos

```python
lista =  'Hola a tod@s'.split(sep = 'a')
lista
```
