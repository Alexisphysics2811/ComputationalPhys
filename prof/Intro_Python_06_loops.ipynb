# Loops

En python existen dos tipos de loops/ciclos, estos son: **for** y **while**. La principal diferencia entre ambos es que el primero emplea ciclos finitos bajo una condición, mientras que el segundo puede emplear tanto ciclos finitos como infinitos. 

### For
Típicamete, los ciclos for se utilizan para iterar sobre estructuras de datos con una dimensión finita, por ejemplo listas, diccionarios, ranges, etc. La sintaxis del ciclo for es la siguiente: 

<center>
    <img src = 'https://misapuntesyacimientos.wordpress.com/wp-content/uploads/2020/06/null-5.jpeg?w=500&h=134'>
</center>

La **variable** (también conocido como **iterador**) es un dato numérico se actualiza en cada ciclo mientras que el **elemento iterable** contiene todos los valores que irá tomando la variable. Veamos un ejemplo

```python
lista_frutas = ['manzana', 'naranja', 'uva', 'pera', 'mango']

for fruta in lista_frutas: # La varible de nombre fruta irá tomando cada uno de los valores guardados en la lista de forma ordenada
    
    texto = 'En la canasta de frutas hay una ' + str(fruta)  # Aquí declaramos la serie de instrucciones que se repetirán en cada ciclo
    print(texto)
```

```python
for i in range(10): # La variable i tomará los valores dentro de range (es decir 0,1,2,3,...,9) 
    print('Este es el ciclo número {}'.format(i+1)) 
```

Dentro de los subloques del ciclo for podemos realizar todo tipo de instrucciones, por ejemplo, hagamos un código que separe número pares e impares dentro de un rango dado por el usuario

```python
# Ejercicio pares e impares
pares = []
impares = []

n = int(input('Ingrese un número entero positivo: '))

for i in range(n): 

    if i%2 == 0:        # Si el residuo del número dividido entre 2 es cero, i.e. es par...
        pares.append(i) # Lo guaradamos en la lista de pares
    
    else:               # caso contrario es impar...
        impares.append(i)

print('Lista de números pares: ', pares)
print('Lista de números impares: ', impares)
```

Otra cosa muy útil es la creación de listas de forma comprimida usando los ciclos for

```python
lista_potencias_cuadradas = [x**2 for x in range(10)] 

print(lista_potencias_cuadradas)
```

y podemos repetir el ejercicio de los pares e impares de forma compacta

```python
n = int(input('Ingrese un número entero positivo: '))

pares = [x for x in range(n) if x%2 == 0]
impares = [x for x in range(n) if x%2 != 0]

print('Lista de números pares: ', pares)
print('Lista de números impares: ', impares)
```

¿Es posible hacer ciclos for infinitos? Sí, pero con un pequeño truco de actualización dentro del ciclo for

```python
lista_frutas = ['manzana', 'naranja', 'uva', 'pera', 'mango']

for fruta in lista_frutas: # La varible de nombre fruta irá tomando cada uno de los valores guardados en la lista de forma ordenada
    
    texto = 'En la canasta de frutas hay una ' + str(fruta)  # Aquí declaramos la serie de instrucciones que se repetirán en cada ciclo
    print(texto)

    if fruta == lista_frutas[-1]:
        respuesta = input('¿Desea agregar una fruta a la canasta? (S/N)').lower() # Convertirmos el string en minúsculas
        
        if respuesta == 's': 
            nueva_fruta = input('Ingrese la fruta: ')
            lista_frutas.append(nueva_fruta)
        else:
            break # ¿Qué es esto de aquí?
    else: 
        continue # ¿Qué es esto de aquí?
```

Cómo vimos, en el ejercicio anterior el ciclo puede ser indefinido siempre y cuando el usuario ingrese frutas de manera continua, con lo cual se cumple nuestro propósito de generar un ciclo infinito. Sin embargo, hay un elemento de código que no hemos explicado qué es y porqué es relevante en este código, esto es, la palabra reservada **continue**.

En python existen palabras reservadas especiales que permiten alterar el flujo de un ciclo, estas son: 

- **continue**: omite el código restante dentro de la iteración actual del bucle y procede inmediatamente a la siguiente iteración.
- **pass**: es una operación nula; no realiza ninguna acción y actúa como marcador de posición. Se utiliza cuando Python requiere sintácticamente una sentencia.
- **break**: esta declaración rompe el ciclo actual y los consecuentes de manera inmediata

En nuestro código usamos continue cuando verificamos que la fruta actual sea igual a la última de la lista, si es el caso, preguntamos si es que se quiere añadir una nueva fruta. Si no se añade ninguna fruta se termina el código de forma tajante

# While

Los ciclos while tienen una sintaxis similar, la diferencia es que no se necesita establecer una estructura iterable sino una condición por verificar. Repitamos el ejercicio de la canaste de frutas, pero ahora en vez de iterar sobre los elementos, lo que haremos sera iterar sobre el índice de los mismos mediante la actualización.

```python
índice = 0

while índice < len(lista_frutas): 

    fruta = lista_frutas[índice] # Extraemos la fruta según el índice
    texto = 'En la canasta de frutas hay una ' + str(fruta)
    print(texto)

    índice += 1 # Es lo mismo que índice = índice + 1
```

Ahora realizaremos el ciclo infinito: 

```python
while True: # Cómo la condición True siempre se cumple, el ciclo no para hasta que cortemos el código directamente del IDE
    print('Ciclo infinito!!')
```

Los ciclos infinitos son peligrosos si no se detienen, para esos caso es muy útil hacer uso de la sentencia brake como cortafuegos de prevención.
